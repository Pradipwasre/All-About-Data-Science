# Categorical Encoding

## 1. When to use what?

### 1 . Label Encoder:

**Use When**

- Binary categories (yes/ no, Male / Female)
- No problem if number are asssigned
- Mostly used for the Target Variable (y).


### 2. One Hot Encoder:

**Use When**:

- Categories have no order
- Color, cities, Product Types, Outlet Types
- Most common encoder in Machiine Learning.


### 3. Ordinal Encoder

**Use When:**

- Categories have a natural order/ranking
- Small < Medium < Large
- low < Medium < High

```python
import pandas as pd

df = pd.DataFrame({
    "Item_fat_Content" : ["Low fat" , "Regular" , "Low fat", "Regular"],
    "Item_Type" : ["Fruits" , "Snacks" , "Diary" , "Fruits"],
    "Outlet_size" : ["Small" , "Medium" , "Large" , "Medium"]
})

print(df)
```

```python
from sklearn.preprocessing import LabelEncoder

df = pd.DataFrame({
    "Item_Fat_Content" : ["Low fat" , "Regular",  "Low fat" , "Regular"]
})

le = LabelEncoder()
df
```

```python
df['Item_Fat_Content'] = le.fit_transform(df['Item_Fat_Content'])
```

```python
df.head()
```

```python
# one hot encoding
df = pd.DataFrame({
    "Item_Type" : ["Fruits" , "Snacks", "Dairy", "Fruits"]
})
df.head()
```

```python
encoded_df = pd.get_dummies(
    df,
    columns= ["Item_Type"],
    dtype = int
)
df
```

```python
encoded_df.head()
```

```python
# Ordinal Encoder

from sklearn.preprocessing import OrdinalEncoder

df = pd.DataFrame({
    "Outlet_Size" : ["small" , "Medium" , "Large" , "Medium"]
})

df.head()
```

```python
oe = OrdinalEncoder(
    categories=[["small" , "Medium" , "Large"]]   
)

df["Size_Encoded"] = oe.fit_transform(df[["Outlet_Size"]])
df.head()
```
